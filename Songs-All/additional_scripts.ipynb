```python
import json

# Specify the path to your album JSON file
input_file_path = 'data/album.json'  # Adjust this path as necessary

def extract_unique_genres(file_path):
    """Extracts and prints unique genres from the album JSON file."""
    unique_genres = set()  # Use a set to collect unique genres

    try:
        with open(file_path, 'r', encoding='utf-8') as input_file:
            # Load the JSON data
            albums = json.load(input_file)

            # Iterate through each album and collect genres
            for album in albums:
                genre = album.get('genre')  # Extract the genre field
                if genre:  # Only add if genre is not None or empty
                    unique_genres.add(genre)

    except FileNotFoundError:
        print(f"File not found: {file_path}")
        return
    except json.JSONDecodeError:
        print("Error decoding JSON. Please check the file format.")
        return

    # Print all unique genres
    print(f"Unique Genres: {len(unique_genres)}")
    for genre in sorted(unique_genres):  # Sort for better readability
        print(genre)

# Run the function to extract and print unique genres
extract_unique_genres(input_file_path)
```

```python
import json

def classify_subgenres(input_file, output_file):
    # Dictionary to hold genres and their corresponding subgenres
    genre_dict = {}

    # Read the text file
    with open(input_file, 'r', encoding='utf-8') as file:
        for line in file:
            # Strip whitespace and split the line into subgenre and genre
            parts = line.strip().split(': ')
            if len(parts) == 2:
                subgenre, genre = parts
                # Add subgenre to the corresponding genre list
                if genre not in genre_dict:
                    genre_dict[genre] = []
                genre_dict[genre].append(subgenre)

    # Write the dictionary to a JSON file
    with open(output_file, 'w', encoding='utf-8') as json_file:
        json.dump(genre_dict, json_file, indent=4)

# Specify input and output file names
input_filename = 'subgenres.txt'  # Change this to your input file name
output_filename = 'data/subgenres_by_genre.json'

# Call the function
classify_subgenres(input_filename, output_filename)

print(f"Subgenres classified by genre have been saved to {output_filename}.")
```
